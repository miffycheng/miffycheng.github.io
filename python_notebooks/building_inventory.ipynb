# Load data

```python
import pandas as pd
import altair as alt

url = "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/building_inventory.csv"
df = pd.read_csv(url)
```

```python
import pandas as pd
import altair as alt


df = df[pd.to_numeric(df["Year Constructed"], errors="coerce").notnull()]
df["Year Constructed"] = df["Year Constructed"].astype(int)
df["Decade"] = (df["Year Constructed"] // 10) * 10


counts = (
    df
    .dropna(subset=["Usage Description"])
    .groupby(["Decade", "Usage Description"], as_index=False)
    .size()
    .rename(columns={"size": "count"})
)

）
click = alt.selection_single(
    fields=["Usage Description"],
    bind="legend",   
    empty="all"      
)

# hover selection
hover = alt.selection_single(
    fields=["Usage Description"],
    on="mouseover",
    empty="all"      
)


chart1 = (
    alt.Chart(counts)
    .transform_filter(click)
    .mark_area()
    .encode(
        x=alt.X("Decade:O", title="Decade Constructed"),
        y=alt.Y("count:Q", title="Number of Buildings"),
        color=alt.Color("Usage Description:N", title="Usage"),

        
        opacity=alt.condition(
            hover,
            alt.value(1),      
            alt.value(0.3)     
        ),

        tooltip=["Decade", "Usage Description", "count"]
    )
    .add_params(hover, click)   
    .properties(
        title="How Building Usage Has Changed Over Decades in Illinois",
        width=600,
        height=350
    )
)

chart1
```

```python
import pandas as pd
import altair as alt

url = "https://raw.githubusercontent.com/UIUC-iSchool-DataViz/is445_data/main/building_inventory.csv"
df = pd.read_csv(url)


df["Square Footage"] = pd.to_numeric(df["Square Footage"], errors="coerce")

size_df = df.dropna(subset=["Usage Description", "Square Footage"])


avg_size = (
    size_df
    .groupby("Usage Description", as_index=False)["Square Footage"]
    .mean()
)


top15 = avg_size.sort_values("Square Footage", ascending=False).head(15)

chart2 = (
    alt.Chart(top15)
    .mark_bar()
    .encode(
        x=alt.X("Square Footage:Q", title="Average Square Footage"),
        y=alt.Y("Usage Description:N",
                sort="-x",
                title="Building Usage")
    )
    .properties(
        title="Average Building Size by Usage (Top 15)",
        width=650,
        height=400
    )
)

chart2
```

```python
# save to html for Jekyll
chart1.save("plot1.html")
chart2.save("plot2.html")
```
